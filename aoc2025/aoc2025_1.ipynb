## Part 1

"The attached document (your puzzle input) contains a sequence of <b>rotations</b>, one per line, which tell you how to open the safe. A rotation starts with an `L` or `R` which indicates whether the rotation should be to the <b>left</b> (toward lower numbers) or to the <b>right</b> (toward higher numbers). Then, the rotation has a <b>distance</b> value which indicates how many clicks the dial should be rotated in that direction....Because the dial is a circle, turning the dial <b>left from `0`</b> one click makes it point at `99`. Similarly, turning the dial <b>right from `99`</b> one click makes it point at `0`....The dial starts by pointing at `50`....The actual password is <b>the number of times the dial is left pointing at `0` after any rotation in the sequence</b>....Analyze the rotations in your attached document. What's the actual password to open the door?" - <a href=https://adventofcode.com/2025/day/1>Eric Wastl, Advent of Code 2025 Day 1</a> (accessed 2025-12-01)

I intend to solve this problem by processing the input file into a list of instructions ("rotations"), proceed through the instructions by simulating them, tally the number of times the dial points at 0, and return the final number after iterating through the full list.

```python
# Preprocess the data

input_file = open(r"C:\Users\EK111\Documents\Fun\Advent of Code\aoc_2025_1_input.txt")
readout = input_file.read()

# We want the data to be in list format for easy comprehension, so we should convert this one big string into a list

readout_as_list = readout.splitlines()

# We also want to make it as easy to find the result as possible, so complex list elements should be split into sub-elements of a list
# element by making the lists nested. Fortunately, that isn't the case here because the L and R instructions can be interpreted as
# positive and negative values.

preprocessed_data = []
for i in range(0, len(readout_as_list)):
    if readout_as_list[i][0] == "L":
        preprocessed_data = preprocessed_data + [-1 * int(readout_as_list[i][1:])]

    else:
        preprocessed_data = preprocessed_data + [int(readout_as_list[i][1:])]
```

```python
# Now we can make a function to process the data:

def aoc2025_1a(data) -> int:
    zero_tally = 0
    dial_value = 50 # Given in the problem statement

    # Iterate through the rotations
    
    for i in range(0, len(data)):
        dial_value = (dial_value + data[i]) % 100

        # Evaluate dial value

        if dial_value == 0:
            zero_tally += 1
            
    return zero_tally
```

```python
# Test the program

aoc2025_1a(preprocessed_data)
```

```python
# Here is a new function to comply with the new requirements:

def aoc2025_1b(data) -> int:
    zero_tally = 0
    dial_value = 50

    for i in range(0, len(data)):
        for j in range(1, abs(data[i]) + 1): # iterate through positive (natural) numbers up to and including the last rotation
            # everything had to be shifted plus 1 because the final rotation is often important
            temp = (dial_value + j if data[i] > 0 else dial_value - j) % 100
            
            if temp == 0:
                zero_tally += 1
            
        dial_value = (dial_value + data[i]) % 100   
        
    return zero_tally
```
