"...<b>find the two entries that sum to</b> `2020` and then multiply those two numbers together....<b>Find the two entries that sum to 2020; what do you get if you multiply them together?</b>" - <a href="https://adventofcode.com/2020/day/1">Eric Wastl, Advent of Code 2020 Day 1</a> (accessed 2025-11-30)

I will approach this problem by creating a function that processes a large numerical input into a list, sort the list into pairs that sum to 2020, and return a list of all products of pairs that sum to 2020.

```python
# My input, downloaded to my device as a TXT file from the URL https://adventofcode.com/2020/day/1/input

aoc20201_input = open(r"C:\Users\EK111\Documents\Fun\Advent of Code\aoc_2020_1_input.txt")
aoc20201_readout = aoc20201_input.read()
```

```python
def aoc2020_1a(raw_data) -> list:
    
    # Turn the readout into a list
    
    expenses = raw_data.splitlines()
    expenses = [int(expenses[i]) for i in range(0, len(expenses))]
    
    # Make an empty list to count viable pairs, and search for them
    
    viable_pairs = []
    
    for i in range(0, len(expenses) - 1):
        for j in range(i + 1, len(expenses)):
            if expenses[i] + expenses[j] == 2020:
                viable_pairs = viable_pairs + [[expenses[i], expenses[j]]]
    
    # Make a response depending on the amount of pairs available

    if len(viable_pairs) < 0:
        print("No pairs found.")
        return [0]

    elif len(viable_pairs) > 2:
        print("There are multiple viable responses.")
        return [viable_pairs[x][0] * viable_pairs[x][1] for x in range(0, len(viable_pairs))]

    else:
        return [viable_pairs[0][0] * viable_pairs[0][1]]
```

```python
# Test the program

aoc2020_1a(aoc20201_readout)
```

"....<b>what is the product of the three entries that sum to `2020`?</b>" - <a href="https://adventofcode.com/2020/day/1">Eric Wastl, Advent of Code 2020 Day 1</a> (accessed 2025-11-30)

My approach now would be to use a three-tier nested list with a conditional inside. It wouldn't be very different.

```python
def aoc2020_1b(raw_data) -> list:
    
    # Turn the readout into a list
    
    expenses = raw_data.splitlines()
    expenses = [int(expenses[i]) for i in range(0, len(expenses))]
    
    # Make an empty list to count viable pairs, and search for them
    
    viable_trios = []
    
    for i in range(0, len(expenses) - 2):
        for j in range(i + 1, len(expenses) - 1):
            for k in range(j + 1, len(expenses)):
                if expenses[i] + expenses[j] + expenses[k] == 2020:
                    viable_trios = viable_trios + [[expenses[i], expenses[j], expenses[k]]]
    
    # Make a response depending on the amount of pairs available

    if len(viable_trios) < 0:
        print("No pairs found.")
        return [0]

    elif len(viable_trios) > 2:
        print("There are multiple viable responses.")
        return [viable_trios[x][0] * viable_trios[x][1] * viable_trios[x][2] for x in range(0, len(viable_trios))]

    else:
        return [viable_trios[0][0] * viable_trios[0][1] * viable_trios[0][2]]
```

```python
# Test the program

aoc2020_1b(aoc20201_readout)
```

## Day 2
"....suppose you have the following list:

`1-3 a: abcde`

`1-3 b: cdefg`

`2-9 c: ccccccccc`

"Each line gives the password policy and then the password. The password policy indicates the lowest and highest number of times a given letter must appear for the password to be valid. For example, `1-3 a` means that the password must contain `a` at least `1` time and at most `3` times....<b>How many passwords are valid</b>...?" - <a href="https://adventofcode.com/2020/day/2">Eric Wastl, Advent of Code 2020 Day 2</a> (accessed 2025-11-30)

My approach to solving this problem is to convert the input (again, a text file) into a nested list where the first element in each sub-list is the minimum character count, the second is the maximum character count, the third is the character required, and the fourth is the applied password. I then assess compliance with each rule and tally up the compliant passwords.

```python
# My input, downloaded to my device as a TXT file from the URL https://adventofcode.com/2020/day/1/input

aoc20202_input = open(r"C:\Users\EK111\Documents\Fun\Advent of Code\aoc_2020_2a_input.txt")
aoc20202_readout = aoc20202_input.read()
```

```python
def aoc2020_2a(raw_data) -> int:
    database = raw_data.splitlines()

    # Separate the data into the nested list I described
    
    database = [
                [
                 int(database[i][0:database[i].find("-")]),
                 int(database[i][(database[i].find("-") + 1):database[i].find(" ")]),
                 database[i][(database[i].find(" ") + 1):database[i].find(":")],
                 database[i][(database[i].find(":") + 2):]
                ] for i in range(0, len(database))
               ]

    # Iterate and tally

    compliant_count = 0

    for i in range(0, len(database)):
        character_count = 0
        
        for j in range(0, len(database[i][3])):
            if database[i][3][j] == database[i][2]:
                character_count += 1

        if character_count >= database[i][0] and character_count <= database[i][1]:
            compliant_count += 1
        
    return compliant_count
```

```python
# Test the program

aoc2020_2a(aoc20202_readout)
```

"....Each policy actually describes two <b>positions in the password</b>, where `1` means the first character, `2` means the second character, and so on....<b>Exactly one of these positions</b> must contain the given letter....How many passwords are valid</b> according to the new interpretation of the policies?" - <a href=https://adventofcode.com/2020/day/2#part2>Eric Wastl, Advent of Code 2020 Day 2</a> (accessed 2025-11-30)

I am going to solve this by changing the conditional algorithm to fit the needs of this new ruleset.

```python
def aoc2020_2b(raw_data) -> int:
    database = raw_data.splitlines()

    # Separate the data into the nested list I described
    
    database = [
                [
                 int(database[i][0:database[i].find("-")]),
                 int(database[i][(database[i].find("-") + 1):database[i].find(" ")]),
                 database[i][(database[i].find(" ") + 1):database[i].find(":")],
                 database[i][(database[i].find(":") + 2):]
                ] for i in range(0, len(database))
               ]

    # Iterate and tally

    compliant_count = 0

    for i in range(0, len(database)):
        hits = (database[i][3][database[i][0] - 1] == database[i][2]) + (database[i][3][database[i][1] - 1] == database[i][2])
        if (hits == 1):
            compliant_count += 1
        
    return compliant_count
```
